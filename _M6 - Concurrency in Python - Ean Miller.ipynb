```python
"""
    Author: Ean Miller
    Program: M6 - Concurrency in Python Ean Miller
    Date: 07/19/2026

13.1  Write the current date as a string to today.txt
13.2  Read the text file today.txt into today_string
13.3  Parse the date from today_string

15.1  Use multiprocessing to create three processes.
      Each waits a random number of seconds (0–1),
      prints the current time, then exits.
"""

# Imports
# -----------------------------
from datetime import date, datetime
import multiprocessing
import time
import random

# 13.1 — Write today's date to today.txt
# -----------------------------
today = date.today().isoformat()  # Example: '2026-07-19'

with open("today.txt", "w") as file:
    file.write(today)

# 13.2 — Read today.txt into today_string
# -----------------------------
with open("today.txt", "r") as file:
    today_string = file.read()

# 13.3 — Parse the date from today_string
# -----------------------------
parsed_date = datetime.strptime(today_string, "%Y-%m-%d").date()

print("Parsed date:", parsed_date)

# 15.1 — Multiprocessing Example
# -----------------------------
def show_time():
    """Wait a random time between 0–1 seconds, then print the current time."""
    wait = random.random()
    time.sleep(wait)
    print(f"Process waited {wait:.3f} seconds. Current time:", datetime.now())

if __name__ == "__main__":
    processes = []

    # Creates and starts 3 processes
    for _ in range(3):
        p = multiprocessing.Process(target=show_time)
        processes.append(p)
        p.start()

    # Waits for all processes to finish
    for p in processes:
        p.join()

    print("All processes completed.")
```
